## 2 biznes sualı üçün null/alternativ hipotezlərin formalaşdırılması

```python
"""
Yanacaq növü (Benzin və Dizel) maşınların satış qiymətinə təsir edirmi?
Sıfır Hipotezi (H0):
Benzin və Dizel mühərrikli maşınların orta satış qiymətləri arasında statistik olaraq əhəmiyyətli fərq yoxdur
Alternativ Hipotez (H1):
Benzin və Dizel mühərrikli maşınların orta satış qiymətləri arasında statistik olaraq əhəmiyyətli fərq var
"""
```

```python
import scipy.stats as stats
benzin_prices = df[df['Yanacaq növü'] == 'Benzin']['price_x'].dropna()
dizel_prices = df[df['Yanacaq növü'] == 'Dizel']['price_x'].dropna()

t_stat, p_value = stats.ttest_ind(benzin_prices, dizel_prices, equal_var=False)
print(f"T-statistikası: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("Result: H0 redd edilir (p < 0.05). Benzin və Dizel maşınların qiymetleri arasında statistik olaraq onemli ferq VAR.")
else:
    print("Result: H0 redd edile bilmir (p >= 0.05). Fərq statistik olaraq vacib DEYİL.")
```

```python
groups = [group['price_x'].dropna() for name, group in df.groupby('Yanacaq növü')]
f_stat, p_value = stats.f_oneway(*groups)
print(f"F-statistikası: {f_stat:.4f}")
print(f"p-value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("Result: H0 rədd edilir (p < 0.05). Yanacaq növləri arasında qiymət fərqi statistik olaraq ƏHƏMİYYƏTLİDİR.")
else:
    print("Result: H0 redd edile bilmir (p >= 0.05). Yanacaq növleri arasında statistik olaraq vacib qiymet ferqi YOXDUR.")
```

## p-dəyərinin və etibar intervalının düzgün interpretasiyası (təkcə "p<0.05 = significant" deyil)
## Parametrik testdən əvvəl fərziyyələrin yoxlanması (normallıq, dispersiya bərabərliyi)

```python
benzin = df[df['Yanacaq növü'] == 'Benzin']['price_x'].dropna()
dizel = df[df['Yanacaq növü'] == 'Dizel']['price_x'].dropna()
stat_benzin, p_benzin = stats.shapiro(benzin.sample(min(len(benzin), 5000)))
stat_dizel, p_dizel = stats.shapiro(dizel.sample(min(len(dizel), 5000)))
print("normalliq testi")
print(f"Benzin p-value: {p_benzin:.5f}")
print(f"Dizel p-value: {p_dizel:.5f}")
```

```python
if p_benzin < 0.05 and p_dizel < 0.05:
    print("Nəticə: Data normal paylanmır (p < 0.05)")
else:
    print("Nəticə: Data normal paylanmaya uyğundur.\n")
```

```python
stat_lev, p_levene = stats.levene(benzin, dizel)
print(f"Levene p-value: {p_levene:.5f}")
if p_levene < 0.05:
    print("Netice: Dispersiyalar BƏRABƏR DEYİL (p < 0.05)")
else:
    print("Netice: Dispersiyalar bərabərdir (p >= 0.05).")
```
