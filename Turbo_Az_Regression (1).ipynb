```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('cars.csv')
```

```python
pd.set_option('display.max_columns',None)
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
df.drop(columns=['id_x','car_rel_url_x','datetime_scrape','datetime_product',
                 'img_url','id_y','cars_id','car_rel_url_y','datetime','car_details_id_x','car_details_id_y','car_rel_url'],inplace=True)
```

```python
df.head(2)
```

```python
df.drop(columns=['price_y', 'currency_x','engine_displacement_num','attributes','engine_displacement_unit',
                 'kilometrage_unit','updated','views','phone','vin','owner_name','shop_name'],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.loc[
    df['Qəzalı'].isna() &
    (df['spare_parts'] == 'Qəzalı'),
    'Qəzalı'
] = 'Bəli'

df.drop(columns=['spare_parts'],inplace=True)
```

```python
df['Model'] = df['Model'].fillna(df['Model'].mode()[0])

df['Vəziyyəti'] = df['Vəziyyəti'].fillna(df['Vəziyyəti'].mode()[0])
```

```python
df['Sahiblər'] = df['Sahiblər'].fillna('Məlum deyil')
```

```python
for col in ['barter','loan','salon','vip','featured']:
    df[col] = df[col].fillna('Xeyr')
```

```python
df['Hansı bazar üçün yığılıb']=df['Hansı bazar üçün yığılıb'].fillna('Məlum deyil')
df['Qəzalı']=df['Qəzalı'].fillna('Məlum deyil')
```

```python
df['Yerlərin sayı'] = df.groupby('Model')['Yerlərin sayı'].transform(lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else 5))
```

```python
df['has_description'] = df['description'].notna().astype(int)
df.drop(columns=['description'], inplace=True)
```

```python
df['extra_info_count'] = df['extra_info'].str.count(r'\*') + 1
df.drop(columns=['extra_info'], inplace=True)
```

```python
rates = {
    'AZN': 1.0,
    'USD': 1.7,
    'EUR': 2.0
}

df['price_azn'] = df['price_x'] * df['currency_y'].map(rates)

df.drop(columns=['city','price_x','name','Buraxılış ili','Yürüş'],inplace=True)
```

```python
df['Yeni'] = df['Yeni'].map({'Xeyr':0, 'Bəli':1})

df['barter'] = (df['barter'] != 'Xeyr').astype(int)
df['loan'] = (df['loan'] != 'Xeyr').astype(int)
```

```python
df['salon'] = (df['salon'] != 'Xeyr').astype(int)
df['vip'] = (df['vip'] != 'Xeyr').astype(int)
df['featured'] = (df['featured'] != 'Xeyr').astype(int)
```

```python
df['day'] = pd.to_datetime(df['day'], format='%d.%m.%Y')
df['month'] = df['day'].dt.month
df['weekday'] = df['day'].dt.weekday
```

```python
df.drop(columns=['day'], inplace=True)
```

```python
df['hour_num'] = df['hour'].str.split(':').str[0].astype(int)
```

```python
df.drop(columns=['hour'], inplace=True)
```

```python
df['Sahiblər'] = df['Sahiblər'].replace({
    'Məlum deyil': -1,
    '0': 0,
    '1': 1,
    '2': 2,
    '3': 3,
    '4 və daha çox': 4
}).astype(int)
```

```python
df['Yerlərin sayı'] = df['Yerlərin sayı'].replace({
    '8+': 8
}).astype(int)
```

```python
parts = df['Mühərrik'].astype(str).str.split('/', expand=True)
```

```python
df['engine_size'] = pd.to_numeric(
    parts[0].str.extract(r'([\d.]+)')[0],
    errors='coerce'
)
```

```python
df['horsepower'] = pd.to_numeric(
    parts[1].str.extract(r'(\d+)')[0],
    errors='coerce'
)
```

```python
df['fuel_type'] = parts[2].str.strip()
df['fuel_type'] = df['fuel_type'].fillna('Məlum deyil')
```

```python
df.drop(columns=['Mühərrik'], inplace=True)
```

```python
df.head()
```

```python
df['fuel_type'].unique()
```

```python
df['fuel_type'] = df['fuel_type'].replace({
    'Hidrogen': 'Digər',
    'Qaz': 'Digər'
})
```

```python
df.drop(columns=['currency_y'], inplace=True)
```

```python
X = df.drop(columns=['price_azn'])
```

```python
y = df['price_azn']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler, OneHotEncoder, LabelEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn import metrics as sm
from warnings import filterwarnings
from sklearn.preprocessing import StandardScaler
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3, random_state = 42)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
```

```python
imp_eng = SimpleImputer(strategy='median')
X_train[['engine_size', 'horsepower']] = imp_eng.fit_transform(X_train[['engine_size', 'horsepower']])
X_test[['engine_size', 'horsepower']] = imp_eng.transform(X_test[['engine_size', 'horsepower']])
```

```python
cat_cols = [
    'Ban növü',
    'Hansı bazar üçün yığılıb',
    'Marka',
    'Model',
    'Qəzalı',
    'Rəng',
    'Sürətlər qutusu',
    'Vəziyyəti',
    'Ötürücü',
    'Şəhər',
    'fuel_type'
]
```

```python
num_cols = [
    'production_year',
    'kilometrage_num',

    'engine_size', 
    'horsepower',

    'Sahiblər',
    'Yerlərin sayı',
    'extra_info_count',

    'month',
    'weekday',
    'hour_num'
]
```

```python
bin_cols=['barter','loan','salon','vip','featured','Yeni','has_description']
```

```python
for col in cat_cols:
    X_train[col] = X_train[col].astype('category')
    X_test[col] = X_test[col].astype('category')
```
