# Datanin yuklenmesi ve ilkin baxis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('googleplaystore.csv')
```

# Data Cleaning

```python
# Reviews sutununun type-ni deyismek
df['Reviews'].unique()
```

```python
df['Reviews'] = df['Reviews'].astype(int)
```

```python
df[df['Reviews'] == '3.0M']
```

```python
df.drop(10472,inplace = True)
```

```python
df['Reviews'] = df['Reviews'].astype(int)
```

```python
# Kilobayt ile texmini hesabla apaririq
df['Size'] = df['Size'].str.replace('M', '000').str.replace('k','').replace('Varies with device',np.nan)
```

```python
df['Size'] = df['Size'].astype(float)
```

```python
df.info()
```

```python
df.head()
```

```python
df['Installs'].value_counts()
```

```python
df['Installs'] = df['Installs'].str.replace('+',"")
```

```python
df['Installs'] = df['Installs'].astype(int)
```

```python
df['Installs'] = df['Installs'].str.replace(',',"")
```

```python
df['Installs'] = df['Installs'].astype(int)
```

```python
df.info()
```

```python
# Using with data column
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
```

```python
#Creation of new columns
df['Day'] = df['Last Updated'].dt.day
df['Month'] = df['Last Updated'].dt.month
df['Year'] = df['Last Updated'].dt.year
```

```python
df.isnull().sum()
```

```python
df['Android Ver'].unique()
```

```python
df['Android Ver'] = df['Android Ver'].str.replace(' and up','').replace('Varies with device',np.nan).str.replace('W','')
```

```python
df['Android Ver'].value_counts()
```

```python
df = df[~df['Android Ver'].str.contains('-', na = False)]
```

```python
df.drop_duplicates(inplace = True)
```

```python
df['Current Ver'].value_counts()
```

```python
df['Current Ver'].replace('Varies with device',np.nan)
```

```python
df['Current Ver'].isna().sum()
```

```python
df[df['Current Ver'] == '5.9.1.1347176.a3ca6ab274']
df.drop(3784,inplace = True)
```

```python
df.info()
```

```python
df.dropna(subset=['Type', 'Current Ver'], inplace=True)
```

```python
df.info()
```

```python
df['Price'].unique()
```

```python
df['Price'] = df['Price'].str.replace('$','')
```

```python
df['Price'] = df['Price'].astype(float)
```

```python
df.info()
```

# Outliers Data-nin boxplot ile teyini

```python
plt.subplot(1,2,1)
sns.boxplot(data = df, y = df['Rating'])
plt.subplot(1,2,2)
sns.boxplot(data = df, y = df['Size'])
plt.tight_layout()
plt.show()
```

# bezi sutunlarda Null deyerlerin doldurulmasi

```python
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Size'] = df['Size'].fillna(df['Size'].median())
```

```python
df['Android Ver'] = df['Android Ver'].fillna(df['Android Ver'].mode()[0])
df.info()
```

# Tekrarlanan Datalarin teyini ve silinmesi

```python
df['App'].duplicated().sum()
```

```python
df.drop_duplicates('App',inplace = True)
```

```python
df.info()
```

# KDE, distribution, histogram ve.s cekilmesi ve IQR hesablanmasi

```python
numeric_columns = [feature for feature in df.columns if df[feature].dtype !='O']
categorical_columns = [feature for feature in df.columns if df[feature].dtype =='O']
```

```python
numeric_columns
```

```python
plt.figure(figsize = (15,8))
for i in range(0,len(numeric_columns)):
  plt.subplot(3,3,i+1)
  sns.kdeplot(x = df[numeric_columns[i]], color = 'b', fill = True)
  plt.xlabel(numeric_columns[i])
  plt.tight_layout()
plt.show()
```

```python
Q1_price = df['Price'].quantile(0.25)
Q3_price = df['Price'].quantile(0.75)
IQR_price = Q3_price - Q1_price
```

```python
lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price
```

```python
df= df[df['Price'] <= upper_price]
```

```python
Q1_size = df['Size'].quantile(0.25)
Q3_size = df['Size'].quantile(0.75)
IQR_size = Q3_size - Q1_size
```

```python
lower_bound_size = Q1_size - 1.5 * IQR_size
upper_bound_size = Q3_size + 1.5 * IQR_size
```

```python
df = df[(df['Size'] >= lower_bound_size) & (df['Size'] <= upper_bound_size)]
```

```python
plt.figure(figsize=(15, 6))
plt.subplot(1, 3, 1)
sns.histplot(df['Rating'], kde=True, color='skyblue')
plt.title('Rating Paylanması (clean)')
```

```python
sns.histplot(df['Size'], kde=True, color='salmon')
plt.title('Size Paylanması (clean)')
plt.show()
```

```python
sns.histplot(df['Price'], kde=True, color='green')
plt.title('Price Paylanması (Təmizlənmiş)')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
corr_matrix = df[['Rating', 'Reviews', 'Size', 'Installs', 'Price']].corr()
sns.heatmap(corr_matrix)
plt.show()
```

```python
plt.figure(figsize=(6, 8))
category_counts = df['Category'].value_counts()
sns.barplot(x=category_counts.values, y=category_counts.index, palette='viridis')
plt.title('Tətbiqlərin Kateqoriyalara Görə Bölgüsü')
plt.xlabel('Tətbiq Sayı')
plt.ylabel('Kateqoriyalar')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Type')
plt.title('Ödənişli və Ödənişsiz Tətbiqlərin Nisbəti')
plt.xlabel('Tip')
plt.ylabel('Sayı')
plt.show()
```
